```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r'F:\statcast-pitch-analysis\Data\statcast_2024_CleanAndMirrored.csv', low_memory=False)
```

```python
fig, ax = plt.subplots(figsize=(10, 8))
sns.scatterplot(
    data=df.sample(50000, random_state=42),  # sample for speed
    x='pfx_x_adj', y='pfx_z',
    hue='pitch_type', alpha=0.6, s=15, ax=ax
)
ax.axhline(0, color='gray', linewidth=0.5)
ax.axvline(0, color='gray', linewidth=0.5)
ax.set_xlabel('Horizontal Movement (ft)')
ax.set_ylabel('Vertical Movement (ft)')
ax.set_title('Pitch Movement Profile: 2024 MLB Season')
plt.savefig(r'F:\statcast-pitch-analysis\Figuresmovement_by_pitch_type.png', dpi=150, bbox_inches='tight')
plt.show()
```
